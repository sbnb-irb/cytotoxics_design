```python
import pandas as pd
import numpy as np
```

## GEx

```python
gex_all_cells = pd.read_csv('../extra_data/GEx_data_from_GDSC.txt', sep='\t')
gex_all_cells.head()
```

```python
metadata = pd.read_csv('../extra_data/metadata_GEx_GDSC.txt', sep='\t')
metadata.head()
```

```python
data2cell = {}
for cll in metadata['Characteristics[cell line]']:
    data = 'DATA.'+metadata[metadata['Characteristics[cell line]'] == cll]['Source Name'].values[0].split('_')[-1]
    data2cell[data] = cll
```

```python
gex_all_cells.columns = ['GENE_SYMBOLS', 'GENE_title']+[data2cell[col] if col in data2cell else col for col in gex_all_cells.columns[2:]]
```

```python
gex_all_cells = gex_all_cells.iloc[:, 2:]  
```

```python
gex_all_cells.head(2)
```

```python
from scipy.spatial.distance import pdist, squareform

# Your subset of interest
subset_cells = ['BxPC-3', 'Capan-2', 'CFPAC-1', 'HPAF-II', 'PANC-10-05', 'SW1990']

# Compute pairwise Euclidean distances between all cell lines
distance_matrix = squareform(pdist(gex_all_cells.T, metric='euclidean'))

# Get background distribution (all pairwise distances)
all_distances = distance_matrix[np.triu_indices_from(distance_matrix, k=1)]

# Identify indices for subset and rest
cell_names = gex_all_cells.columns
subset_indices = [i for i, cell in enumerate(cell_names) if cell in subset_cells]
rest_indices = [i for i, cell in enumerate(cell_names) if cell not in subset_cells]

# Compute distances within the subset
subset_distances = distance_matrix[np.ix_(subset_indices, subset_indices)]
subset_distances = subset_distances[np.triu_indices_from(subset_distances, k=1)]

# Compute distances between subset and the rest
subset_vs_rest_distances = distance_matrix[np.ix_(subset_indices, rest_indices)].flatten()

# Print summary
print(f"Background distances: {len(all_distances)} pairs")
print(f"Subset distances: {len(subset_distances)} pairs")
print(f"Subset vs. Rest distances (closest only): {len(subset_vs_rest_distances)} values")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.rc('font', family='sans-serif')
# Create a DataFrame for easier plotting
plot_data = pd.DataFrame({
    'Distance': np.concatenate([all_distances, subset_distances, subset_vs_rest_distances]),
    'Category': (['All'] * len(all_distances)) + 
                (['PC'] * len(subset_distances)) + 
                (['PC vs. rest'] * len(subset_vs_rest_distances))
})

# Plot the distributions
plt.figure(figsize=(4.5, 3))
custom_palette = {'All':'#5A72B5',
                  'PC':'#EA5A49',
                  'PC vs. rest':'#7CAF2A'
}
sns.kdeplot(data=plot_data, x='Distance', hue='Category', fill=True, common_norm=False, alpha=0.4,zorder=2, palette=custom_palette)

# Labels and title
plt.xlabel('Euclidean distance', fontsize=12)
plt.ylabel('Density', fontsize=12)
plt.title('Gene Expression (1,018 cell lines)')
# plt.legend(title='Distance Type')
plt.grid(ls='--',color='lightgrey', zorder=0)

# Show the plot
plt.show()
```
