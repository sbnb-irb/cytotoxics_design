```python
from rdkit.Chem import AllChem, Draw
from rdkit.DataStructs import TanimotoSimilarity

for e in combinations:
    pos_cll = e[0]
    neg_cll = e[1]
    try:
        smiles_A = [
            cpd2smiles_irb[s] for s in cpds_that_worked_IRB[pos_cll + "_vs_" + neg_cll]
        ]
    except:
        print(pos_cll, neg_cll)
        continue

    cpd_id = cpds_that_worked_IRB[pos_cll + "_vs_" + neg_cll]
    smiles_B = confirmation.SMILES[
        confirmation["positives " + e[0] + " vs " + e[1]] == 1
    ].values

    # Convert SMILES to RDKit molecule objects and compute fingerprints
    mols_A = [
        Chem.MolFromInchi(Chem.MolToInchi(Chem.MolFromSmiles(smile)))
        for smile in smiles_A
    ]
    mols_B = [
        Chem.MolFromInchi(Chem.MolToInchi(Chem.MolFromSmiles(smile)))
        for smile in smiles_B
    ]
    fps_A = [AllChem.GetMorganFingerprintAsBitVect(mol, 2) for mol in mols_A]
    fps_B = [AllChem.GetMorganFingerprintAsBitVect(mol, 2) for mol in mols_B]

    # Compute the maximum Tanimoto similarity for each molecule in A
    results = []
    for i, fp_a in enumerate(fps_A):
        max_sim = 0
        closest_idx = -1
        for j, fp_b in enumerate(fps_B):
            sim = TanimotoSimilarity(fp_a, fp_b)
            if sim > max_sim:
                max_sim = sim
                closest_idx = j
        results.append((mols_A[i], mols_B[closest_idx], max_sim))
        print(smiles_A[i])
        print(np.array(smiles_B)[closest_idx])
    # Plot the structures and their similarity scores
    n = len(results)
    fig, axes = plt.subplots(n, 1, figsize=(8, 4 * n))
    if n == 1:
        axes = [axes]  # Ensure axes is a list when n=1

    for i, (ax, (mol_a, mol_b, sim)) in enumerate(zip(axes, results)):
        # Create a combined image of the two molecules
        img = Draw.MolsToImage([mol_a, mol_b], subImgSize=(500, 500))
        ax.imshow(img)
        ax.axis("off")
        ax.set_title(f"Tanimoto Similarity: {sim:.2f}, (cpd %s)" % cpd_id[i])

    plt.tight_layout()
    # plt.savefig(
    #     "%s_%s_molecules_IRB.svg" % (e[0], e[1]),
    #     format="svg",
    #     dpi=300,
    #     bbox_inches="tight",
    # )
    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from scipy.optimize import curve_fit 
```

```python
# Code to fit the logistic curve
def custom_logistic(x, log_ic50, hill_slope):
    # X is in log scale
    return 100 / (1 + 10 ** (hill_slope * (log_ic50 - x)))
```

```python
clls_, cpds_, ic50s_ = [],[],[]
for exercise in list(exercises_generated_cpds):
    pos_cll = exercise[0]
    neg_cll = exercise[1]
    for cpd in exercises_generated_cpds[exercise]:
        df = data[data["Sample"] == cpd]

        # Log-transform the concentrations (log10 scale)
        df["LogConc"] = np.log10(
            df["Conc (uM)"] * 1e-6
        )  # Convert μM to M before log10 transformation
        x_data = df["LogConc"].values  # Log-transformed concentrations
        plt.figure(figsize=(4, 3))
        for cll in [pos_cll, neg_cll]:

            df["Mean_%s" % cll] = df[["%s_Rep1" % cll, "%s_Rep2" % cll]].mean(axis=1)
            df["StdDev_%s" % cll] = df[["%s_Rep1" % cll, "%s_Rep2" % cll]].std(axis=1)
            # Prepare data for fitting

            y_data = df["Mean_%s" % cll].values  # % Inhibition

            # Curve fitting
            p0 = [0, 1]  # Initial guesses for logIC50 and HillSlope
            try:
                popt, pcov = curve_fit(
                    custom_logistic, list(x_data) + [-11], list(y_data) + [-3], p0=p0
                )
            except:
                print(f"Failed to fit {cpd} for {cll}")
                plt.errorbar(
                    x_data,
                    y_data,
                    yerr=df["StdDev_%s" % cll],
                    fmt="o",
                    color=colors[cll],
                    capsize=3,
                    zorder=2,
                    markersize=4,
                    elinewidth=0.8,
                )
                label = cll2name[cll] + "\nIC50 = Unstable"
                plt.plot(
                    x_smooth,
                    [df["Mean_%s" % cll].mean()] * len(x_smooth),
                    label=label,
                    color=colors[cll],
                    zorder=2,
                )
                continue

            # Extract parameters
            log_ic50, hill_slope = popt
            molar_concentration = 10**log_ic50
            # Convert Molar (M) to micromolar (uM)
            ic50_micromolar = molar_concentration * 1e6
            print(cpd)
            print(cll)
            print(
                f"Fitted Parameters:\nIC50 (uM): {ic50_micromolar}, HillSlope: {hill_slope}"
            )

            # Generate smooth x values for plotting
            x_smooth = np.linspace(min(x_data), max(x_data), 100)
            y_fitted = custom_logistic(x_smooth, log_ic50, hill_slope)

            # Plot the data with error bars and the fitted curve
            plt.errorbar(
                x_data,
                y_data,
                yerr=df["StdDev_%s" % cll],
                fmt="o",
                color=colors[cll],
                capsize=3,
                zorder=2,
                markersize=4,
                elinewidth=0.8,
            )
            if ic50_micromolar > 100:
                label = cll2name[cll] + "\nIC50 = Unstable"
            else:
                label = cll2name[cll] + "\nIC50 = %.2f uM" % ic50_micromolar
            plt.plot(x_smooth, y_fitted, label=label, color=colors[cll], zorder=2)

            clls_.append(cll)
            cpds_.append(cpd)
            ic50s_.append(ic50_micromolar)

        # Set x-axis labels as log(M)
        tick_values = np.linspace(-8, -4, 5)  # Define tick positions
        tick_labels = [f"{val:.0f}" for val in tick_values]  # Format tick labels
        plt.xticks(tick_values, tick_labels)

        plt.xlabel("Log[Drug] M", fontsize=11)
        plt.ylabel("% Inhibition", fontsize=11)
        plt.title(cpd)
        plt.legend(fontsize=11)
        plt.grid(ls="--", color="lightgrey", zorder=0)
        plt.ylim(-25, 110)
        plt.savefig(
            "./DR_curves/%s_%s_%s.svg" % (cpd, pos_cll, neg_cll),
            dpi=300,
            format="svg",
            bbox_inches="tight",
        )
        plt.show()
        plt.close()
```
